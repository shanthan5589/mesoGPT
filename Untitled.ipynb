```python
from pathlib import Path

import matplotlib.pyplot as plt
from matplotlib.patches import Patch


# ---------------------------------------------------------------------
# Configuration
# ---------------------------------------------------------------------

OUTPUT_DIR = Path("experiments/exp_003/assets")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

plt.rcParams.update(
    {
        "figure.dpi": 120,
        "savefig.dpi": 200,
        "font.size": 11,
        "axes.titlesize": 13,
        "axes.labelsize": 11,
        "axes.spines.top": False,
        "axes.spines.right": False,
    }
)

BASELINE_COLOR = "#6B7280"
BEST_COLOR = "#059669"
SDPA_COLORS = [
    "#059669",
    "#3B82F6",
    "#8B5CF6",
]

CONFIGS = [
    "Explicit\nMB 16",
    "SDPA\nMB 16",
    "SDPA\nMB 32",
    "SDPA\nMB 64",
]

COLORS = [
    BASELINE_COLOR,
    SDPA_COLORS[0],
    SDPA_COLORS[1],
    SDPA_COLORS[2],
]


# ---------------------------------------------------------------------
# Experiment results
# ---------------------------------------------------------------------

UPDATE_TIME = [
    17.16,
    4.20,
    4.71,
    5.11,
]

THROUGHPUT = [
    30_553,
    124_831,
    111_314,
    102_600,
]

TRAINING_ONLY_MFU = [
    4.94,
    20.20,
    18.01,
    16.60,
]

PEAK_MEMORY_MIB = [
    27_969,
    12_457,
    21_715,
    42_369,
]


# ---------------------------------------------------------------------
# Helper functions
# ---------------------------------------------------------------------

def save_figure(fig, filename):
    output_path = OUTPUT_DIR / filename

    fig.savefig(
        output_path,
        bbox_inches="tight",
        facecolor="white",
    )

    plt.show()
    plt.close(fig)

    print(f"Saved: {output_path}")


def label_bars(ax, bars, formatter):
    for bar in bars:
        value = bar.get_height()

        ax.annotate(
            formatter(value),
            xy=(
                bar.get_x() + bar.get_width() / 2,
                value,
            ),
            xytext=(0, 5),
            textcoords="offset points",
            ha="center",
            va="bottom",
            fontsize=9,
        )


# ---------------------------------------------------------------------
# Graph 1: Explicit Attention vs. PyTorch SDPA
# ---------------------------------------------------------------------

def plot_performance_overview():
    fig, axes = plt.subplots(
        2,
        2,
        figsize=(13, 9),
    )

    axes = axes.flatten()

    panels = [
        {
            "values": UPDATE_TIME,
            "title": "Optimizer-update time",
            "ylabel": "Seconds per optimizer update",
            "formatter": lambda value: f"{value:.2f}s",
            "direction": "Lower is better",
        },
        {
            "values": THROUGHPUT,
            "title": "Training throughput",
            "ylabel": "Tokens per second",
            "formatter": lambda value: f"{value:,.0f}",
            "direction": "Higher is better",
        },
        {
            "values": TRAINING_ONLY_MFU,
            "title": "Training-only MFU",
            "ylabel": "MFU (%)",
            "formatter": lambda value: f"{value:.2f}%",
            "direction": "Higher is better",
        },
        {
            "values": PEAK_MEMORY_MIB,
            "title": "Peak GPU memory",
            "ylabel": "Memory (MiB)",
            "formatter": lambda value: f"{value:,.0f}",
            "direction": "Lower is better",
        },
    ]

    for ax, panel in zip(axes, panels):
        values = panel["values"]

        bars = ax.bar(
            CONFIGS,
            values,
            color=COLORS,
            width=0.68,
        )

        ax.set_title(panel["title"])
        ax.set_ylabel(panel["ylabel"])
        ax.set_ylim(0, max(values) * 1.20)
        ax.grid(axis="y", alpha=0.25)
        ax.set_axisbelow(True)

        label_bars(
            ax,
            bars,
            panel["formatter"],
        )

        ax.text(
            0.98,
            0.96,
            panel["direction"],
            transform=ax.transAxes,
            ha="right",
            va="top",
            fontsize=9,
            color="#4B5563",
        )

    axes[2].text(
        0.02,
        0.96,
        "Evaluation and end-to-end overhead excluded",
        transform=axes[2].transAxes,
        ha="left",
        va="top",
        fontsize=8.5,
        color="#4B5563",
    )

    fig.legend(
        handles=[
            Patch(
                color=BASELINE_COLOR,
                label="Explicit-attention baseline",
            ),
            Patch(
                color=BEST_COLOR,
                label="Best result: SDPA, MB 16",
            ),
            Patch(
                color=SDPA_COLORS[1],
                label="Other SDPA configurations",
            ),
        ],
        loc="lower center",
        ncol=3,
        frameon=False,
    )

    fig.suptitle(
        "Explicit Attention vs. Flash Attention",
        fontsize=17,
        fontweight="bold",
    )

    fig.text(
        0.5,
        0.935,
        "Global batch size fixed at 512 sequences",
        ha="center",
        color="#4B5563",
    )

    fig.subplots_adjust(
        top=0.88,
        bottom=0.13,
        hspace=0.35,
        wspace=0.25,
    )

    save_figure(
        fig,
        "performance_overview.png",
    )


# ---------------------------------------------------------------------
# Graph 2: Throughput speedup and relative peak memory
# ---------------------------------------------------------------------

def plot_relative_improvement():
    labels = [
        "SDPA\nMB 16",
        "SDPA\nMB 32",
        "SDPA\nMB 64",
    ]

    positions = list(range(len(labels)))

    throughput_speedup = [
        value / THROUGHPUT[0]
        for value in THROUGHPUT[1:]
    ]

    memory_percent = [
        value / PEAK_MEMORY_MIB[0] * 100
        for value in PEAK_MEMORY_MIB[1:]
    ]

    fig, axes = plt.subplots(
        1,
        2,
        figsize=(13, 5.5),
    )

    # Left: throughput speedup
    throughput_bars = axes[0].bar(
        positions,
        throughput_speedup,
        color=SDPA_COLORS,
        width=0.65,
    )

    axes[0].axhline(
        1,
        color=BASELINE_COLOR,
        linestyle="--",
        linewidth=1.5,
        label="Explicit-attention baseline: 1×",
    )

    axes[0].set_title("Training throughput")
    axes[0].set_ylabel("Improvement over baseline (×)")
    axes[0].set_xticks(positions)
    axes[0].set_xticklabels(labels)
    axes[0].set_ylim(
        0,
        max(throughput_speedup) * 1.20,
    )
    axes[0].grid(axis="y", alpha=0.25)
    axes[0].set_axisbelow(True)
    axes[0].legend(frameon=False)

    label_bars(
        axes[0],
        throughput_bars,
        lambda value: f"{value:.2f}×",
    )

    # Right: peak memory relative to baseline
    memory_bars = axes[1].bar(
        positions,
        memory_percent,
        color=SDPA_COLORS,
        width=0.65,
    )

    axes[1].axhline(
        100,
        color=BASELINE_COLOR,
        linestyle="--",
        linewidth=1.5,
        label="Explicit-attention baseline: 100%",
    )

    axes[1].set_title("Peak memory relative to baseline")
    axes[1].set_ylabel("Baseline memory (%)")
    axes[1].set_xticks(positions)
    axes[1].set_xticklabels(labels)
    axes[1].set_ylim(
        0,
        max(memory_percent) * 1.20,
    )
    axes[1].grid(axis="y", alpha=0.25)
    axes[1].set_axisbelow(True)
    axes[1].legend(frameon=False)

    label_bars(
        axes[1],
        memory_bars,
        lambda value: f"{value:.1f}%",
    )

    fig.suptitle(
        "Relative Improvement observed after using Flash Attention",
        fontsize=16,
        fontweight="bold",
    )

    fig.tight_layout(
        rect=(0, 0, 1, 0.92),
    )

    save_figure(
        fig,
        "relative_improvement.png",
    )


# ---------------------------------------------------------------------
# Generate both graphs
# ---------------------------------------------------------------------

if __name__ == "__main__":
    plot_performance_overview()
    plot_relative_improvement()
```
